```python
import numpy as np
import matplotlib.pyplot as plt

def plot_res(vp_data, vs_data, rho_data, xticks=None):
    plt.figure(figsize=(12, 4))
    xlist = np.arange(len(vp_data))

    # Plot the curves
    plt.plot(xlist, vp_data, marker='o', c='r', label='vp')
    plt.plot(xlist, vs_data, marker='v', c='g', label='vs')
    plt.plot(xlist, rho_data, marker='*', c='b', label='rho')

    # Mark minimum values
    for data, label, color, marker in zip([vp_data, vs_data, rho_data], 
                                          ['vp', 'vs', 'rho'], 
                                          ['r', 'g', 'b'], 
                                          ['o', 'v', '*']):
        min_idx = np.argmin(data)
        min_val = data[min_idx]
        plt.scatter(min_idx, min_val, c=color, s=100, edgecolors='k', zorder=5)
        plt.annotate(f"min={min_val:.2f}", 
                     (min_idx, min_val), 
                     textcoords="offset points", 
                     xytext=(0,10), 
                     ha='center', 
                     fontsize=9, 
                     color=color)

    plt.legend()
    if xticks is not None:
        plt.xticks(xlist, xticks)
    plt.xlabel("Index")
    plt.ylabel("Value")
    plt.title("Curves with Minimum Value Annotations")
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```
