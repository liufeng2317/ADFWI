```python
import numpy as np
import matplotlib.pyplot as plt

def plot_res(vp_data, vs_data, rho_data, xticks=None):
    plt.figure(figsize=(12, 4))
    xlist = np.arange(len(vp_data))

    # Plot the curves
    plt.plot(xlist, vp_data, marker='o', c='r', label='vp')
    plt.plot(xlist, vs_data, marker='v', c='g', label='vs')
    plt.plot(xlist, rho_data, marker='*', c='b', label='rho')

    # Mark minimum values
    for data, label, color, marker in zip([vp_data, vs_data, rho_data], 
                                          ['vp', 'vs', 'rho'], 
                                          ['r', 'g', 'b'], 
                                          ['o', 'v', '*']):
        min_idx = np.argmin(data)
        min_val = data[min_idx]
        plt.scatter(min_idx, min_val, c=color, s=100, edgecolors='k', zorder=5)
        plt.annotate(f"min={min_val:.2f}", 
                     (min_idx, min_val), 
                     textcoords="offset points", 
                     xytext=(0,10), 
                     ha='center', 
                     fontsize=9, 
                     color=color)

    plt.legend()
    if xticks is not None:
        plt.xticks(xlist, xticks)
    plt.xlabel("Index")
    plt.ylabel("Value")
    plt.title("Curves with Minimum Value Annotations")
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

```python
from scipy.interpolate import griddata
import matplotlib.transforms as mtransforms
    
def plot_vel_single_for_all(fig,ax,v,title="",MSE="",vmin=None,vmax=None,text_color='w'):
    plt.rc('font',family='Times New Roman')
    # cmap = "bwr"
    cmap = "rainbow"
    # plm = ax.pcolormesh(x_mesh, z_mesh, v,cmap=cmap,vmin=vmin,vmax=vmax)
    x = np.arange(nx*3)*dx/3/1000
    z = np.arange(nz*3)*dz/3/1000
    x_mesh_new,z_mesh_new = np.meshgrid(x,z)
    v_new = griddata((x_mesh.flatten(), z_mesh.flatten()), v.flatten(), (x_mesh_new, z_mesh_new), method='cubic')
    
    plm = ax.pcolormesh(x_mesh_new, z_mesh_new, v_new,cmap=cmap,vmin=vmin,vmax=vmax,shading="nearest")
    ax.invert_yaxis()
    ax.tick_params(labelsize = 16)
    ax.set_title(title,fontsize=16)
    ax.text(0.1,0.35,MSE,fontsize=16,c=text_color)
    return plm

def add_right_cax(ax, pad, width, shrink=1):
    height = ax.get_position().y1 - ax.get_position().y0
    bottom_position = ax.get_position().y0 + height * (1 - shrink) / 2
    axpos = ax.get_position()
    caxpos = mtransforms.Bbox.from_extents(
        axpos.x1 + pad,
        bottom_position,
        axpos.x1 + pad + width,
        bottom_position + height * shrink
    )
    cax = ax.figure.add_axes(caxpos)

    return cax

import matplotlib as mpl
def add_bottom_cax(ax, pad, height,shrink=1):
    axpos = ax.get_position()
    width = axpos.x1 - axpos.x0
    left_position = axpos.x0 + width * (1 - shrink) / 2
    caxpos = mpl.transforms.Bbox.from_extents(
        left_position,
        axpos.y0 - pad,
        left_position + width * shrink,
        axpos.y0 - pad + height
    )
    cax = ax.figure.add_axes(caxpos)
    return cax

def plot_vel_singleline_for_all(ax,v_true,v_init,v_inv,x_distance,title,show_xlabel=True,show_ylabel=False,show_legend=False):
    ax.plot(v_true[:,int(x_distance//dx)]/1000,  z, c='k',   linewidth=2, linestyle="-" ,label="True")
    ax.plot(v_init[:,int(x_distance//dx)]/1000,  z, c='gray',linewidth=2, linestyle="-" ,label="Init")
    ax.plot(v_inv [:,int(x_distance//dx)]/1000,  z, c='r',   linewidth=2, linestyle="--",label="Inverted")
    ax.tick_params(labelsize = 16)
    if not show_xlabel:
        ax.set_xticks([])
    # else:
    #     ax.tick_params(labelsize = 15)
        
    if not show_ylabel:
        ax.set_yticks([])
    else:
        ax.tick_params(labelsize = 12)
    ax.invert_yaxis()
    
    if show_legend:
        ax.legend(fontsize = 12)
    ax.set_title(title,fontsize=16)
```
