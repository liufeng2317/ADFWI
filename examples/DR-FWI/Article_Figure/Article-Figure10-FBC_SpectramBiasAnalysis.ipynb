```python
from matplotlib.patches import Circle

def plot_vel_single_for_all(fig,ax,v,title="",MSE="",vmin=None,vmax=None, cmap = "rainbow"):
    x = np.arange(nx*3)*dx/3/1000
    z = np.arange(nz*3)*dz/3/1000
    x_mesh_new,z_mesh_new = np.meshgrid(x,z)
    from scipy.interpolate import griddata
    v_new = griddata((x_mesh.flatten(), z_mesh.flatten()), v.flatten(), (x_mesh_new, z_mesh_new), method='cubic')
    
    plm = ax.pcolormesh(x_mesh_new, z_mesh_new, v_new,cmap=cmap,vmin=vmin,vmax=vmax,shading="nearest")
    ax.invert_yaxis()
    ax.tick_params(labelsize = 16)
    ax.set_title(title,fontsize=16)
    ax.text(0.2,0.35,MSE,fontsize=16,c="k")
    return plm

def plot_spectral_single_for_all(fig,ax,v_spec,vmin=None,vmax=None, cmap = "viridis",radius_ratio = 4):
    dx_km = dx / 1000 
    dz_km = dz / 1000
    fx = np.fft.fftfreq(nx, d=dx_km)
    fz = np.fft.fftfreq(nz, d=dz_km)
    fx_shifted = np.fft.fftshift(fx)
    fz_shifted = np.fft.fftshift(fz)
    FX, FZ = np.meshgrid(fx_shifted, fz_shifted)
    plm = ax.pcolormesh(FX, FZ, v_spec, cmap=cmap, vmin=vmin, vmax=vmax,shading="nearest")
    radius_freq = (min(nx, nz) // radius_ratio) * np.abs(fx[1] - fx[0])  # 转换为频率单位
    circle = Circle((0, 0), radius_freq, color='k', linestyle='--', fill=False, linewidth=1)
    ax.add_patch(circle)
    ax.tick_params(labelsize = 16)
    return plm

import matplotlib.transforms as mtransforms
def add_right_cax(ax, pad, width):
    axpos = ax.get_position()
    caxpos = mtransforms.Bbox.from_extents(
        axpos.x1 + pad,
        axpos.y0,
        axpos.x1 + pad + width,
        axpos.y1
    )
    cax = ax.figure.add_axes(caxpos)

    return cax

def add_bottom_cax(ax, pad, height):
    import matplotlib.transforms as mtransforms
    
    axpos = ax.get_position()
    caxpos = mtransforms.Bbox.from_extents(
        axpos.x0,
        axpos.y0 - pad - height,
        axpos.x1,
        axpos.y0 - pad
    )
    cax = ax.figure.add_axes(caxpos)
    
    return cax

def plot_vel_singleline_for_all(ax,v_true,v_init,v_inv,x_distance,title,show_xlabel=True,show_ylabel=False,show_legend=False):
    ax.plot(v_true[:,int(x_distance//dx)]/1000,  z, c='k',   linewidth=2, linestyle="-" ,label="True")
    ax.plot(v_init[:,int(x_distance//dx)]/1000,  z, c='gray',linewidth=2, linestyle="-" ,label="Init")
    ax.plot(v_inv [:,int(x_distance//dx)]/1000,  z, c='r',   linewidth=2, linestyle="--",label="Inverted")
    ax.tick_params(labelsize = 16)
    if not show_xlabel:
        ax.set_xticks([])

    if not show_ylabel:
        ax.set_yticks([])
    else:
        ax.tick_params(labelsize = 12)
    ax.invert_yaxis()
    
    if show_legend:
        ax.legend(fontsize = 12)
    ax.set_title(title,fontsize=16)
```

```python
import numpy as np

def compute_circular_statistic_2d(true_model, pred_model, size=0.2):
    """
    Compute circular frequency statistics between two 2D velocity models.
    The function evaluates the average amplitude ratio of the predicted and 
    true models in the frequency domain as a function of normalized radius.

    Parameters
    ----------
    true_model : np.ndarray
        Ground truth 2D velocity model (nz x nx)
    pred_model : np.ndarray
        Predicted 2D velocity model
    size : float
        Normalized step size (0 < size < 1), controlling 
        the number of circular frequency bands

    Returns
    -------
    avg_ratio_list : list
        Mean amplitude ratio in each circular band
    radii_norm : list
        Normalized frequency radii corresponding to each band
    """

    # --- Input validation ---
    assert pred_model.ndim == 2 and true_model.ndim == 2, \
        "Inputs must be 2D velocity models."
    assert 0 < size < 1, "Parameter 'size' must be in (0,1)."
    assert pred_model.shape == true_model.shape, \
        "Predicted and true models must have the same shape."

    # --- FFT magnitude (center-shifted) ---
    ft_pred = np.abs(np.fft.fftshift(np.fft.fft2(pred_model)))
    ft_true = np.abs(np.fft.fftshift(np.fft.fft2(true_model)))

    # --- Amplitude ratio (normalized) ---
    ratio = ft_pred / (ft_true + 1e-8)
    ratio = np.clip(ratio, 0, 1)

    # --- Prepare circular masks ---
    h, w = ratio.shape
    center = (w // 2, h // 2)
    Y, X = np.ogrid[:h, :w]
    dist = np.sqrt((X - center[0])**2 + (Y - center[1])**2)
    max_r = np.max(dist)

    # --- Circular averaging ---
    avg_ratio_list = []
    radii_norm = []

    pre_mask = np.zeros_like(ratio, dtype=np.int32)
    pre_sz = 0.0  # previous radius (relative)

    # bands = [0.05,0.15,0.25,0.4,0.6,1]
    # bands = [0.1,0.20,0.30,0.5,0.8,1]

    for sz in np.linspace(size, 1, int(1 / size)):
    # for sz in bands:
        radius = max_r * sz
        mask = (dist <= radius).astype(np.int32)
        ring_mask = mask - pre_mask
        pre_mask = mask

        if np.sum(ring_mask) > 0:
            avg_ratio = np.sum(np.abs(ratio * ring_mask)) / np.sum(ring_mask)
            avg_ratio_list.append(avg_ratio)
            radii_norm.append((pre_sz, sz))  # 返回区间
            pre_sz = sz

    return avg_ratio_list, radii_norm


def compute_iterative_circular_statistic(true_model, model_series, size=0.2):
    """
    Analyze the frequency-domain evolution of iterative inversion models.

    This function computes the circular spectral amplitude ratio 
    between each iterative velocity model and the true model, based 
    on the same normalized frequency radii.

    Parameters
    ----------
    true_model : np.ndarray
        Ground truth 2D velocity model (nz x nx)
    model_series : np.ndarray
        Predicted velocity models across iterations (N_iter x nz x nx)
    size : float
        Normalized step size (0 < size < 1), controlling 
        the number of frequency bands

    Returns
    -------
    ratios_all : np.ndarray
        Array of shape [N_iter, N_band], containing amplitude ratios 
        for each iteration
    radii_norm : np.ndarray
        Normalized frequency radii shared by all iterations
    """

    # --- Input validation ---
    model_series = np.asarray(model_series)
    assert model_series.ndim == 3, "model_series must have shape [N_iter, H, W]"
    assert true_model.ndim == 2, "true_model must be 2D"
    assert model_series.shape[1:] == true_model.shape, \
        "Predicted models must match the true model shape."

    n_iter = model_series.shape[0]

    # --- Compute reference radii using the first model ---
    from copy import deepcopy
    first_model = deepcopy(model_series[0])
    avg_ratio_ref, radii_norm = compute_circular_statistic_2d(true_model, first_model, size=size)
    n_band = len(avg_ratio_ref)

    # --- Initialize result container ---
    ratios_all = np.zeros((n_iter, n_band), dtype=np.float32)
    ratios_all[0] = avg_ratio_ref

    # --- Loop over all models ---
    for i in range(1, n_iter):
        avg_ratio, _ = compute_circular_statistic_2d(true_model, model_series[i], size=size)
        ratios_all[i] = avg_ratio

    return ratios_all, np.array(radii_norm)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# === FFT magnitude (center-shifted) ===
ft_true = np.abs(np.fft.fftshift(np.fft.fft2(true_v)))

# === Wave number axes (1/m) ===
ny, nx = true_v.shape
kx = np.fft.fftshift(np.fft.fftfreq(nx, d=dx))   # cycles per meter
ky = np.fft.fftshift(np.fft.fftfreq(ny, d=dz))
# 如果希望使用弧度波数 (rad/m)，取消下两行注释
# kx = 2 * np.pi * kx
# ky = 2 * np.pi * ky

# === Prepare meshgrid for plotting ===
X, Z = np.meshgrid(kx, ky)
R = np.sqrt(X**2 + Z**2)

# === Plot FFT magnitude (log scale recommended) ===
plt.figure(figsize=(10, 5.5))
plt.pcolormesh(X, Z, np.log10(ft_true + 1e-12), cmap='turbo')
plt.colorbar(label='log10(FFT magnitude)')

# === Add N concentric circles (relative spacing mode) ===
size = 0.15  # relative spacing ratio (e.g., 0.15 means 15% per ring)
r_max = np.sqrt(kx.max()**2 + ky.max()**2)

radii_rel = np.linspace(size, 1, int(1 / size))
radii_0 = np.concatenate(([0], radii_rel)) * r_max
radii = radii_rel * r_max

theta = np.linspace(0, 2 * np.pi, 400)
alphas = np.linspace(0.4, 0.1, len(radii))

for i, r in enumerate(radii):
    plt.plot(r * np.cos(theta), r * np.sin(theta), 'w--', lw=1.2, alpha=0.8)
    mask = (R >= radii_0[i]) & (R < radii_0[i + 1])
    plt.contourf(
        X, Z, mask,
        levels=[0.5, 1],
        colors='red',
        alpha=alphas[i]
    )
    if i == len(radii) - 1:
        break
    r_text = r - (radii_0[i + 1] - radii_0[i]) / 2
    theta_text = np.pi / 4
    plt.text(r_text * np.cos(theta_text), r_text * np.sin(theta_text), f'B({i+1})',
             fontsize=13, ha='center', va='center', color='black', fontweight='bold')

plt.xlim(kx.min(), kx.max())
plt.ylim(ky.min(), ky.max())

plt.xlabel(r'Wavenumber ($m^{-1}$)', fontsize=15)
plt.ylabel(r'Wavenumber ($m^{-1}$)', fontsize=15)
plt.tick_params(labelsize=15)

# plt.savefig("./Figures/Figure10a_Frequency_Band_Definition.png", dpi=300, bbox_inches='tight')
# plt.savefig("./Figures_PDF/Figure10a_Frequency_Band_Definition.pdf", dpi=300, bbox_inches='tight',format="pdf")
plt.savefig("./Figures_SVG/Figure10a_Frequency_Band_Definition.svg", dpi=300, bbox_inches='tight',format="svg")
plt.show()
```
